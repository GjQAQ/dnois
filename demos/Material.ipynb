```python
def show_curve(m):
    min_wl = dnois.Length.as_default(m.min_wl, m.default_unit) if m.min_wl != 0. else 380e-9
    max_wl = dnois.Length.as_default(m.max_wl, m.default_unit) if m.max_wl != float('inf') else 760e-9
    wl = torch.linspace(min_wl, max_wl, 100)
    fig, ax = plt.subplots()
    plt.grid()
    ax.plot(wl, m.n(wl))
    ax.set_xlabel(r'$\lambda/m$')
    ax.set_ylabel(r'$n$')
```

```python
show_curve(dnois.mt.get('air'))
```

```python
show_curve(dnois.mt.get('silica'))
```

```python
NOA61 = dnois.mt.Cauchy('NOA61', 1.5375, 8.29045e-3, 2.11046e-4, 0.45, 1.55, 'um')

show_curve(NOA61)
```
