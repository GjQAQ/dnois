```python
import matplotlib.pyplot as plt
import torch

import dnois
```

```python
def show_curve(m):
    wl = torch.linspace(m.min_wl, m.max_wl, 100)
    fig, ax= plt.subplots()
    plt.grid()
    ax.plot(wl, m.n(wl))
    ax.set_xlabel(r'$\lambda/\mu m$')
    ax.set_ylabel(r'$n$')
```

```python
SiO_2 = dnois.mt.Sellmeier1(
    'SiO_2',
    [0.6961663, 0.4079426, 0.8974794],
    [0.0684043 ** 2, 0.1162414 ** 2, 9.896161 ** 2],
    0.21, 6.7, 'um'
)

show_curve(SiO_2)
```

```python
NOA61 = dnois.mt.Cauchy('NOA61', 1.5375, 8.29045e-3, 2.11046e-4, 0.45, 1.55, 'um')

show_curve(NOA61)
```

```python
BK7 = dnois.mt.Sellmeier1(
    'BK7',
    [1.03961212, 0.231792344, 1.01046945],
    [0.00600069867, 0.0200179144, 103.560653],
    0.3, 2.5, 'um'
)

show_curve(BK7)
```
